```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.ensemble import BaggingClassifier,RandomForestClassifier
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
```

```python
X,y = make_classification(n_features=5, n_redundant=0, n_informative=5,n_clusters_per_class=1)
```

```python
# here, we created dataframe 5 input columns and one target column

df = pd.DataFrame(X,columns=['col1','col2','col3','col4','col5'])
df['target'] = y
print(df.shape)
df.head()
```

```python
# created a object of baggingclassifier with only 2 columns
bag = BaggingClassifier(max_features=2)
```

```python

bag.fit(df.iloc[:,:5],df.iloc[:,-1])
```

```python
# Here, we can see that bagging classifier only uses two column for sampling, they are x[0] and x[1]
plt.figure(figsize=(12,12))
plot_tree(bag.estimators_[0])
plt.show()
```

```python
## created a object of baggingclassifier with only 2 column
rf = RandomForestClassifier(max_features=2)
```

```python
rf.fit(df.iloc[:,:5],df.iloc[:,-1])
```

```python
# here, we can see that random forest classifier uses column1 for first node, then uses column 2,3,4, to split at each node by using sampling at each node
# therefor, it increase randomness
plt.figure(figsize=(12,12))
plot_tree(rf.estimators_[4])
plt.show()
```
